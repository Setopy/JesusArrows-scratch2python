# Mission 3: Guess my number
**You will learn:** `random`, `while` loops, `if`, and turning words into numbers with `int()`

**In Python:**
```python
import random
secret = random.randint(1, 10)
guess = 0
while guess != secret:
    guess = int(input("Guess 1 to 10: "))
    if guess < secret:
        print("Too low")
    if guess > secret:
        print("Too high")
print("You got it!")
```
⚠️ **Python flips the question.** Scratch says *repeat **until** guess = secret*. Python says *keep going **while** guess **!=** secret*. (`!=` means "is not equal to".)

💡 Whatever you type into `input` is **writing**, even a number. `int()` turns it into a real number.

# Mission 5: Turtle race
**You will learn:** two sprites, random numbers, `and`, and `if` / `else`

Two turtles race across the stage, and each one moves a random amount every turn. In Scratch, each sprite would have its own script. In Python, **one loop moves both turtles**.

```python
from ColabTurtlePlus.Turtle import *
import random
clearscreen()
setup(480, 360)

red = Turtle()
red.shape("turtle")
red.color("red")
red.penup()
red.goto(-200, 50)
red.pendown()

blue = Turtle()
blue.shape("turtle")
blue.color("blue")
blue.penup()
blue.goto(-200, -50)
blue.pendown()

while red.xcor() < 200 and blue.xcor() < 200:
    red.forward(random.randint(1, 20))
    blue.forward(random.randint(1, 20))

if red.xcor() > blue.xcor():
    print("Red wins!")
else:
    print("Blue wins!")
```
💡 The word before the dot says **which sprite** to move: `red.forward(10)` moves only the red turtle.
